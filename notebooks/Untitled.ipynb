```python
import re, collections

path = "elite_kaggle_corpus_clean_filtered.txt"
counts = collections.Counter()
bar_re = re.compile(r'^\[BAR\]\s*(.*?)\s*\[RHY=', re.IGNORECASE)

with open(path, "r", encoding="utf-8", errors="ignore") as f:
    for line in f:
        m = bar_re.search(line)
        if not m:
            continue
        bar = m.group(1).strip()
        if not bar:
            continue
        words = re.findall(r"[a-zA-Z']+", bar)
        n = len(words)
        if n < 3 or n > 30:
            continue
        counts[n] += 1

print(counts.most_common(10))
```
